# Exercise: General Bayesian statistics Tarot

The first Tarot reading of the course: form pairs and answer at least
one question with the help of the Tarot deck, taking into account the
meanings listed below.

See [the introduction](tarot.md) for more about Tarot reading and how to
do these exercises.

## Suit cards

Pentacles: Measurement Swords: Information Cups: Variation Wands:
Computation

Numbers mean amounts.

Face cards mean: - Page = zero - Knight = more - Queen = the right
amount - King = infinity
