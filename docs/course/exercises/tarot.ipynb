## What is Tarot reading?

[Tarot reading](https://en.wikipedia.org/wiki/Tarot_card_reading) is a
form of fortune telling involving a special deck of playing cards with
evocative illustrations
([here](https://en.wikipedia.org/wiki/Rider%E2%80%93Waite_Tarot) is the
most famous set).

The deck involves four suits called wands, cups, swords and pentacles
(these are alternative names for clubs, hearts, spades and diamonds) and
22 extra face cards called “major arcana”.

In a typical reading, a questioner, possibly with help from a reader,
thinks of a question, chooses some cards then arranges them face down in
a pattern called a “spread”, with the positions in the pattern
indicating some meaning. For example, a three card spread might
represent the past, present and future. They then turn the cards over
and use them as a prompt to think or talk about the answer to their
question.

There are no rules for how to interpret the cards, but there is no
shortage of starting points. First, there is a lot going on in the
pictures: you can just ask how the things in the picture make you feel
and how they relate to your question and spread. For more inspiration
there is a large amount of material online about the cards’ many-layered
mystical meanings: [this card
list](https://www.dailytarotdraw.com/tarot-card-list) is a good starting
point. You may want to use the convention that a card’s orientation
(i.e. whether it is right-way-up or upside-down) indicates whether to
focus on good or bad connotations.

## How to do the exercise

First, find a partner and decide who will be the questioner and who will
be the reader.

Then decide a question to discuss. Any question works: open-ended
questions leave a lot of room for interpretation, but it can also be fun
to ask extremely specific ones!
[Here](https://www.sherylwagnermedium.com/blog/questions-to-ask-tarot-cards)
are some
[lists](https://www.tarotmasterguide.com/blogs/tarot-insights/what-to-ask-the-tarot-50-powerful-questions-for-self-discovery)
of
[questions](https://forum.thetarot.guru/t/49-of-the-best-tarot-questions/521)
for inspiration.

Now choose a spread. Feel free to
[research](https://www.tarotnova.app/blog/most-used-tarot-spreads), but
in my opinion the three card
[past-present-future](https://daily-tarot-girl.com/tarot-card-spreads/3-card-tarot-spread-past-present-future/)
spread is a good default.

Now conduct the reading. Lay out the cards face down in your chosen
spread, then turn them over one by one and discuss the answer to your
question as it emerges! The only twist on a standard Tarot reading is
that, as well as its other meanings and connotations, in this exercise
each card represents a learning concept, which you can find on the page
for the exercise. Try to incorporate the concepts into your discussion:
it’s fine to look them up on the internet.
